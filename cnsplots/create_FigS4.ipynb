```python
# read mapping file for keywords to sub-, mid-, and top-level categories
keywords_map = pd.read_csv("./input_data/Category_map.csv") 

# create a mapping from subcategories to top-level categories
cat_to_top = keywords_map[[COLNAME_MAP, TOP_COLNAME_MAP]].drop_duplicates().set_index(COLNAME_MAP)[TOP_COLNAME_MAP].to_dict()
```

```python
# read in dataframe
preds = pd.read_csv("./input_data/Preds_BacVir95_Pharokka_Phold.tsv", sep="\t")
preds = preds[preds["assigned_subcat_initial"] == "Unknown"].sort_values(["P(Predicted)"], ascending=False)

sns.set_theme(style="whitegrid", palette="viridis")

counts = preds["Predicted Class"].value_counts().reset_index()
counts.columns = ["Predicted Class", "Frequency"]
counts["Top_Cat"] = counts["Predicted Class"].map(cat_to_top)

plt.figure(figsize=(6.5, 15))

sns.barplot(data=counts, x="Frequency", y="Predicted Class", errorbar=None, hue="Top_Cat", palette=topcat_colors_adjusted, dodge=False, edgecolor="black")

plt.xlabel("Frequency", fontsize=18)
plt.ylabel("Predicted Class", fontsize=18)

ax = plt.gca()

for spine in ax.spines.values():
    spine.set_visible(True)
    spine.set_color("black")
    spine.set_linewidth(1)

plt.legend(
    title="Top Category",
    bbox_to_anchor=(1, 0),
    loc="lower right"
)

plt.grid(axis="x", alpha=0.75)

# Save final figure
outdir = Path("./plots/")
outdir.mkdir(parents=True, exist_ok=True)
plt.savefig(outdir / "FigureS4.svg", dpi=300, bbox_inches="tight")
plt.show()
```
