```python
import pandas as pd
import numpy as np
from sklearn.metrics import precision_recall_fscore_support
```

```python
# high-confidence predictions for the unknome
all_preds = pd.read_csv("../cnsplots/input_data/Preds_BacVir95_Pharokka_Phold.tsv", sep="\t")
reps95 = pd.read_csv("../cnsplots/input_data/all_new_representatives_assigned_cat_0.95.tsv", sep="\t")
reps95_phage = reps95[reps95["table_origin"] == "phg"]
print(len(reps95_phage))

def clean_df(df, colname):
    """"Helper function to clean the dataframe by replacing entries with semicolons in the specified column with 'Unknown'."""
    mask = df[colname].str.contains(";")
    df.loc[mask, colname] = "Unknown"
    return df
    
all_preds = clean_df(all_preds, "assigned_subcat_initial")
preds95 = all_preds[all_preds["protein_ID"].isin(reps95_phage["protein_ID"])]
print(len(preds95))

# report for varying confidence thresholds
conf_thresholds = [0.9, 0.8, 0.7, 0.6, 0.5, 0.4, 0.3, 0.2, 0.1, 0]
#N_unknome = len(all_preds[all_preds["assigned_subcat_initial"] == "Unknown"])
N_unknome = len(preds95[preds95["assigned_subcat_initial"] == "Unknown"])
print(f"Unlabeled proteins in 95% dataset: {N_unknome}")
for t in conf_thresholds:
    subset = preds95[preds95["P(Predicted)"] >= t]
    unknown_subset = subset[(subset["assigned_subcat_initial"] == "Unknown")].sort_values(["P(Predicted)"], ascending=False)
    total_pred = len(unknown_subset)

    subset_phold = unknown_subset[unknown_subset["score_weighted"] >= t]
    subset_valid_phold = subset_phold[(subset_phold["assigned_subcat_phold"] != "Unknown") & (subset_phold["assigned_subcat_phold"] != "Other")]
    total_phold = len(subset_valid_phold)

    print(f"Threshold: {t}")
    print(f"Number of predictions | Percentage:         {total_pred} | {total_pred / N_unknome:.2f}")
    print(f"Number of Phold annotations | Percentage:   {total_phold} | {total_phold / N_unknome:.2f}")
```

```python
# same for BacVir20
all_preds = pd.read_csv("../cnsplots/input_data/Preds_BacVir20_Pharokka_Phold.tsv", sep="\t")
reps20 = pd.read_csv("../cnsplots/input_data/all_new_representatives_assigned_cat_0.2.tsv", sep="\t")
reps20_phage = reps20[reps20["table_origin"] == "phg"]
print(len(reps20_phage))

all_preds = clean_df(all_preds, "assigned_subcat_initial")
preds20 = all_preds[all_preds["protein_ID"].isin(reps20_phage["protein_ID"])]
print(len(preds20))

# report for varying confidence thresholds
conf_thresholds = [0.9, 0.8, 0.7, 0.6, 0.5, 0.4, 0.3, 0.2, 0.1, 0]
#N_unknome = len(all_preds[all_preds["assigned_subcat_initial"] == "Unknown"])
N_unknome = len(preds20[preds20["assigned_subcat_initial"] == "Unknown"])
print(f"Unlabeled protein in 20% dataset: {N_unknome}")
for t in conf_thresholds:
    subset = preds20[preds20["P(Predicted)"] >= t]
    unknown_subset = subset[(subset["assigned_subcat_initial"] == "Unknown")].sort_values(["P(Predicted)"], ascending=False)
    total_pred = len(unknown_subset)

    subset_phold = unknown_subset[unknown_subset["score_weighted"] >= t]
    subset_valid_phold = subset_phold[(subset_phold["assigned_subcat_phold"] != "Unknown") & (subset_phold["assigned_subcat_phold"] != "Other")]
    total_phold = len(subset_valid_phold)

    print(f"Threshold: {t}")
    print(f"Number of predictions | Percentage:         {total_pred} | {total_pred / N_unknome:.2f}")
    print(f"Number of Phold annotations | Percentage:   {total_phold} | {total_phold / N_unknome:.2f}")
```

```python
# get stats for BASEL phages
basel_95 = pd.read_csv("../cnsplots/input_data/BASEL_BacVir95_final.tsv", sep="\t")

# Pharokka
prop_assigned_pharokka = len(basel_95.loc[(basel_95["assigned_subcat_pharokka"] != "Unknown") & (basel_95["assigned_subcat_pharokka"] != "Other")]) / len(basel_95) * 100
print(f"Pharokka assigned {prop_assigned_pharokka}")

# Phold
prop_assigned_phold = len(basel_95.loc[(basel_95["assigned_subcat_phold"] != "Unknown") & (basel_95["assigned_subcat_phold"] != "Other")]) / len(basel_95) * 100
print(f"Phold assigned {prop_assigned_phold}")

# Pharokka-Phold
prop_assigned_pharokka_phold = len(basel_95.loc[
    (basel_95["assigned_subcat_pharokka"] != "Unknown") & (basel_95["assigned_subcat_pharokka"] != "Other") |
    (basel_95["assigned_subcat_phold"] != "Unknown") & (basel_95["assigned_subcat_phold"] != "Other") |
    ((basel_95["assigned_subcat_pharokka"] != "Unknown") & (basel_95["assigned_subcat_pharokka"] != "Other") & (basel_95["assigned_subcat_phold"] != "Unknown") & (basel_95["assigned_subcat_phold"] != "Other"))
    ]) / len(basel_95) * 100
print(f"Pharokka-Phold assigned {prop_assigned_pharokka_phold}")

# Classifier at 50% confidence
classifier50 = (len(basel_95[basel_95["P(Predicted)"] > 0.5])/len(basel_95)) * 100
print(f"Classifier (BacVir95) assigned {classifier50} at 50% confidence")

# same for BacVir20
basel_20 = pd.read_csv("../cnsplots/input_data/BASEL_BacVir20_final.tsv", sep="\t")
classifier50_20 = (len(basel_20[basel_20["P(Predicted)"] > 0.5])/len(basel_20)) * 100
print(f"Classifier (BacVir20) assigned {classifier50_20} at 50% confidence")
```

```python
# Precision, Recall, F1 (BacVir95)
basel_95 = clean_df(basel_95, "assigned_subcat_initial")
basel_95_known = basel_95[basel_95["assigned_subcat_initial"] != "Unknown"]

# calculate metrics
weighted_precision, weighted_recall, weighted_f1, _ = precision_recall_fscore_support(basel_95_known["assigned_subcat_initial"], basel_95_known["Predicted Class"], average="weighted", zero_division=0)
macro_precision, macro_recall, macro_f1, _ = precision_recall_fscore_support(basel_95_known["assigned_subcat_initial"], basel_95_known["Predicted Class"], average="macro", zero_division=0)

print(f"Weighted -  Precision: {weighted_precision:.2f}  Recall: {weighted_recall:.2f}   F1: {weighted_f1:.2f}")
print(f"Macro -     Precision: {macro_precision:.2f}  Recall: {macro_recall:.2f}   F1: {macro_f1:.2f}")
```
