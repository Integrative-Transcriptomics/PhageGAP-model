```python
import pandas as pd
```

```python
# 95% sequence identity dataset
df = pd.read_csv("../cnsplots/input_data/Preds_BacVir95_Pharokka_Phold.tsv", sep="\t")

def clean_df(df, colname):
    """"Helper function to clean the dataframe by replacing entries with semicolons in the specified column with 'Unknown'."""
    mask = df[colname].str.contains(";")
    df.loc[mask, colname] = "Unknown"
    return df
    
df = clean_df(df, "assigned_subcat_initial")

reps95 = pd.read_csv("../cnsplots/input_data/all_new_representatives_assigned_cat_0.95.tsv", sep="\t")
reps95_phage = reps95[reps95["table_origin"] == "phg"]

df95 = df[df["protein_ID"].isin(reps95_phage["protein_ID"])]
unknome95 = df95[df95["assigned_subcat_initial"] == "Unknown"]
print(len(unknome95))
unknome95.to_csv("TableS4_unknome_95.csv", index=False)


# 20% sequence identity dataset
df = pd.read_csv("../cnsplots/input_data/Preds_BacVir20_Pharokka_Phold.tsv", sep="\t")
df = clean_df(df, "assigned_subcat_initial")

reps20 = pd.read_csv("../cnsplots/input_data/all_new_representatives_assigned_cat_0.2.tsv", sep="\t")
reps20_phage = reps20[reps20["table_origin"] == "phg"]

df20 = df[df["protein_ID"].isin(reps20_phage["protein_ID"])]
unknome20 = df20[df20["assigned_subcat_initial"] == "Unknown"]
print(len(unknome20))
unknome20.to_csv("TableS4_unknome_20.csv", index=False)
```
