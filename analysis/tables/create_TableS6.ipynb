```python
import pandas as pd
```

```python
df = pd.read_csv("../cnsplots/input_data/all_new_representatives_assigned_cat_0.95.tsv", sep="\t")
def clean_df(df, colname):
    """"Helper function to clean the dataframe by replacing entries with semicolons in the specified column with 'Unknown'."""
    mask = df[colname].str.contains(";")
    df.loc[mask, colname] = "Unknown"
    return df
    
df = clean_df(df, "assigned_subcat")
known = df[df["assigned_subcat"] != "Unknown"]
print(len(known))
known.to_csv("TableS6_AugmentedDataset.csv", index=False)
```
